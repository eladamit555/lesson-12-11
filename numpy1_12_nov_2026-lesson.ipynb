```python
np.random.seed(42)
np1 = np.random.randint(0, 101, (6,5))
print(np1)
print()

print('max', np1.max())
print('min', np1.min())
print('argmax', np1.argmax())
print('argmin', np1.argmin())
print('shape', np1.shape)
print()

def get_row_col(index, matrix):
    return matrix // matrix.shape[1], index % matrix[0]
print('max index 11 location', get_row_col(np1.argmax(), np1)) 
print('min index 11 location', get_row_col(np1.argmin(), np1)) 

```

### targil 4

```python
np1 = np.arange(5, 25 + 1)
print(np1)
print(np1[np1 >= 20])
print(np1[np1 % 5 == 0])
print(np1[(np1 >= 20) & (np1 % 2 == 0)])
print(np1[(np1 < 10) | (np1 > 20)])
```

```python
np1 = np.arange(0, 10 + 1)
print(np1)
print('mean', np1.mean())
print('median', np.median(np1))
print('std', np1.std())
print('sum', np1.sum())
```

```python
np1 = np.arange(0, 24 + 1).reshape(5,5)
print(np1)
print()

print('mean', np1.mean())
print('sum', np1.sum())
print()

print('column sum')
print(np1.sum(axis=0))
print()

print('row sum axis=1')
print(np1.sum(axis=1).reshape(5,1))
```
